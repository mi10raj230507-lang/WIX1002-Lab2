```python
#1.Create a numpy array of size 2x2
import numpy as nmp
array_2x2 = nmp.array([[1,2],
                       [3,4]])
print(array_2x2)


```

```python
#2.Create a random numpy array of size 3x4
random_array=nmp.random.rand(3,4)
print(random_array)
```

```python
#3.Create a numpy array of ones and a numpy array of zeroes
ones_array = nmp.ones((4,4))
zeroes_array = nmp.zeros((4,4))

print("Array of ones:")
print(ones_array)

print("Array of zeroes:")
print(zeroes_array)
```

```python
#4.Create a random numpy array of size 6x3 and transpose it
random_array = nmp.random.rand(6,3)
print("Original array:")
print(random_array)
transposed_array = random_array.T

print("Transposed array:")
print(transposed_array)
```

```python
#5.Create two random numpy array of size n x p and p x m and thencompute the dot product
A=nmp.random.rand(3,4)
B=nmp.random.rand(4,5)

print("Matrix A:")
print(A)

print("Matrix B:")
print(B)

DP=nmp.dot(A,B)
print("Dot Product")
print(DP)
```
